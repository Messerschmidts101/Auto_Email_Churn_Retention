```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain.embeddings.sentence_transformer import SentenceTransformerEmbeddings
from langchain_community.document_loaders import TextLoader
from langchain_community.document_loaders import DirectoryLoader
from langchain_community.vectorstores import Chroma
import os

strContextKnowledgeDirectory = 'LLM_Database'
objEmbeddingModel = SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")
objLoader = DirectoryLoader(strContextKnowledgeDirectory, glob="**/*.txt", loader_cls=TextLoader, show_progress=False)
raw_documents = objLoader.load()
text_splitter = RecursiveCharacterTextSplitter(chunk_size=700, chunk_overlap=100)
documents = text_splitter.split_documents(raw_documents)
objVectorStore = Chroma.from_documents(
    documents, 
    objEmbeddingModel, 
    persist_directory = strContextKnowledgeDirectory
)
```

# 1. Query Documents Based On Query

```python
strQuestion = "Tell me about segment 1"
objDocuments = objVectorStore.similarity_search(strQuestion, k=2)

for doc in objDocuments:
    print(doc.page_content)
```

# 2. Get Vector Embeddings Of Your Query

```python
strQuestion = "Tell me about segment 1"
lisfltEmbeddings = objEmbeddingModel.embed_query(strQuestion)
print(lisfltEmbeddings)  # a list of floats (e.g. 384-dimension vector)
```

# 3. Get Docs Based On Vector

```python
objDocuments = objVectorStore.similarity_search_by_vector(lisfltEmbeddings, k=3)

for doc in objDocuments:
    print(doc.page_content)
```
